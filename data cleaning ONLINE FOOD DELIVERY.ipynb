```python
import pandas as pd

df = pd.read_csv("ONLINE_FOOD_DELIVERY_ANALYSIS.csv")

# Missing values count for each column
print(df.isnull().sum())
#dtypes
print(df.dtypes)
```

```python
#Median for Numerical Columns
num_cols = [
    'Customer_Age',
    'Delivery_Time_Min',
    'Distance_km',
    'Order_Value',
    'Discount_Applied',
    'Final_Amount',
    'Delivery_Rating'
]

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())


#Categorical Columns
cat_cols = [
    'Customer_Gender',
    'City',
    'Area',
    'Cuisine_Type',
    'Payment_Mode',
    'Peak_Hour'
]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

#Cancellation Reason
df['Cancellation_Reason'] = df['Cancellation_Reason'].fillna('Not Cancelled')

#order date and order time
df['Order_Date'] = df['Order_Date'].fillna(df['Order_Date'].mode()[0])

df['Order_Time'] = df['Order_Time'].fillna(df['Order_Time'].mode()[0])

#save null cleaned dataset
df.to_csv("ONLINE_FOOD_DELIVERY_CLEANED.csv", index=False)

#read cleaned dataset
df = pd.read_csv("ONLINE_FOOD_DELIVERY_CLEANED.csv")


#verify null values handled dataset
print(df.isnull().sum())


# Outlier detection by IQR
Q1 = df['Delivery_Time_Min'].quantile(0.25)
Q3 = df['Delivery_Time_Min'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)

#min max value of Delivery_Time_Min
print(df['Delivery_Time_Min'].describe())
print(df['Delivery_Time_Min'].min())
print(df['Delivery_Time_Min'].max())

#count outliers
outliers = df[
    (df['Delivery_Time_Min'] < -102.5) |
    (df['Delivery_Time_Min'] > 325.5)
]

print("Number of outliers:", len(outliers))



#correcting invalid values
# Restaurant ratings greater than 5
print("Invalid Restaurant Ratings:")
print(df[df['Restaurant_Rating'] > 5].shape[0])

# Delivery ratings greater than 5
print("Invalid Delivery Ratings:")
print(df[df['Delivery_Rating'] > 5].shape[0])

# Negative profit margins
print("Negative Profit Margins:")
print(df[df['Profit_Margin'] < 0].shape[0])

df = df[df['Restaurant_Rating'] <= 5]
print("Restaurant Rating > 5:", (df['Restaurant_Rating'] > 5).sum())
#check rows in restaurant rating
print("Rows after cleaning:", df.shape)

```

```python
# Outlier detection by IQR
Q1 = df['Order_Value'].quantile(0.25)
Q3 = df['Order_Value'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
#min max value of Order_Value
print(df['Order_Value'].describe())
print(df['Order_Value'].min())
print(df['Order_Value'].max())

#count outliers
outliers = df[
    (df['Order_Value'] < -1777.0 ) |
    (df['Order_Value'] > 5455.0)
]

print("Number of outliers:", len(outliers))
```

```python
import pandas as pd
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=np.number).columns

print("Outlier Summary (IQR Method)\n")

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\nColumn: {col}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of Outliers: {len(outliers)}")
```
